Task 1: Create the records

```python
records = [
  {"id": 1, "plan": "Basic", "usage": 12.5, "churned": "No"},
  {"id": 2, "plan": "Premium", "usage": None, "churned": "Yes"},
  {"id": 3, "plan": "Standard", "usage": 8.0, "churned": "No"},
  {"id": 4, "plan": "Basic", "usage": 15.0, "churned": "Yes"},
  {"id": 5, "plan": "Premium", "usage": 22.5, "churned": "No"},
  {"id": 6, "plan": "Standard", "usage": None, "churned": "No"},
  {"id": 7, "plan": "Basic", "usage": 9.5, "churned": "Yes"},
  {"id": 8, "plan": "Premium", "usage": 20.0, "churned": "No"},
]
```

Task 2: Check the data

```python
print("CHECK THE DATA")
missing = 0
total = 0
for r in records:
  if r["usage"] is None:
    missing += 1
  else:
    total += r["usage"]
mean_usage = round(total / (len(records) - missing), 2)
print("Missing usage values:", missing)
print("Mean usage (valid values):", mean_usage)
```

Task 3: Fill the gaps (imputation)

```python
for r in records:
  if r["usage"] is None:
    r["usage"] = mean_usage

print("CLEANED DATA")
print(f"{'ID':<4}{'Plan':<10}{'Usage':>7}{'Churned':>9}")
for r in records:
  print(f"{r['id']:<4}{r['plan']:<10}{r['usage']:>7.2f}{r['churned']:>9}")
```

Task 4: Encode the categories

```python
plan_codes = {"Basic": 0, "Standard": 1, "Premium": 2}
churn_codes = {"No": 0, "Yes": 1}
for r in records:
  r["plan_code"] = plan_codes[r["plan"]]
  r["churn_flag"] = churn_codes[r["churned"]]

print("ENCODED DATA")
print(f"{'ID':<4}{'plan_code':>10}{'usage':>8}{'churn_flag':>12}")
for r in records:
  print(f"{r['id']:<4}{r['plan_code']:>10}{r['usage']:>8.2f}{r['churn_flag']:>12}")
```

Task 5: Churn rate by plan

```python
churned_total = 0
by_plan = {}
for r in records:
  churned_total += r["churn_flag"]
  if r["plan"] not in by_plan:
    by_plan[r["plan"]] = [0, 0]
  by_plan[r["plan"]][0] += 1
  by_plan[r["plan"]][1] += r["churn_flag"]

print(f"Overall churn rate: {churned_total / len(records) * 100:.1f}%")
print("CHURN RATE BY PLAN")
for plan in sorted(by_plan):
  count, lost = by_plan[plan]
  print(f"{plan:<10}{lost}/{count} {lost / count * 100:>5.1f}%")
```
