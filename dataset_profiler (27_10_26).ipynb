Task 1: Create the dataset

```python
data = {
  "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
  "age": [21, 22, None, 20, 23, 21, 22, 24],
  "hours": [5.5, 3.0, 7.0, 2.5, 6.0, 4.5, None, 8.0],
  "city": ["Kochi", "Kollam", "Kochi", "Thrissur", "Kollam", "Kochi", "Kollam", "Kochi"],
  "score": [78, 65, 88, 52, 74, 70, 69, 91],
}
```

Task 2: Shape of the data

```python
rows = len(data["student_id"])
print("DATASET PROFILE")
print("=" * 40)
print("Rows:", rows)
print("Columns:", len(data))
print("Column names:", ", ".join(data))
```

Task 3: Missing values per column

```python
print("MISSING VALUES")
for col, values in data.items():
  missing = 0
  for v in values:
    if v is None:
      missing += 1
  print(f"{col:<12}{missing:>3}")
```

```python
print("\nMISSING VALUES")
for ky,val in data.items():
  print(f'{ky:<12} {val.count(None):>3}')
```

Task 4: Numeric summary

```python
print("NUMERIC SUMMARY")
print(f"{'Column':<8}{'Count':>6}{'Mean':>8}{'Min':>7}{'Max':>7}")
for col in ["age", "hours", "score"]:
  valid = []
  for v in data[col]:
    if v is not None:
      valid.append(v)
  print(f"{col:<8}{len(valid):>6}{sum(valid) / len(valid):>8.2f}{min(valid):>7}{max(valid):>7}")
```

Task 5: Category frequency

```python
print("CITY FREQUENCY")
city_counts = {}
for city in data["city"]:
  city_counts[city] = city_counts.get(city, 0) + 1
for city, n in city_counts.items():
  print(f"{city:<10}{n:>2} {'#' * n}")
```
