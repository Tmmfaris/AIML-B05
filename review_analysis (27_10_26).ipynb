Task 1: Create the data

```python
reviews = [
  "The delivery was fast and the product is great",
  "Terrible service and the product arrived broken",
  "Good quality but the delivery was late",
  "I love this product and the support team is excellent",
  "Poor packaging and slow delivery",
  "Great product and fast delivery very happy",
]
stop_words = ["the", "and", "is", "was", "but", "i", "this", "very", "a"]
lexicon = {"good": 1, "great": 2, "excellent": 3, "love": 2, "fast": 1, "happy": 2, "slow": -1, "late": -1, "poor": -2, "bad": -2, "broken": -2, "terrible": -3}
```

Task 2: Top 5 words

```python
counts = {}
for review in reviews:
  for word in review.lower().split():
    if word not in stop_words:
      counts[word] = counts.get(word, 0) + 1
pairs = []
for word, n in counts.items():
  pairs.append((n, word))
pairs.sort(reverse=True)
print("TOP 5 WORDS")
print("-" * 20)
for n, word in pairs[:5]:
  print(f"{word:<12}{n:>3}")
```

Task 3: Sentiment score of each review

```python
print("SENTIMENT PER REVIEW")
print("-" * 20)
scores = []
labels = []
for i, review in enumerate(reviews, start=1):
  score = 0
  for word in review.lower().split():
    score += lexicon.get(word, 0)
  if score > 0:
    label = "Positive"
  elif score < 0:
    label = "Negative"
  else:
    label = "Neutral"
  scores.append(score)
  labels.append(label)
  print(f"Review {i}: {score:>3} {label}")
```

Task 4: Summary

```python
label_count = {}
for label in labels:
  label_count[label] = label_count.get(label, 0) + 1

print("SUMMARY")
for label in sorted(label_count):
  print(f"{label:<10}{label_count[label]}")
print(f"Average score: {sum(scores) / len(scores):.2f}")
```

Challenge

```python
positive_words = {}
negative_words = {}
for i in range(len(reviews)):
  for word in reviews[i].lower().split():
    if word in stop_words:
      continue
    if labels[i] == "Positive":
      positive_words[word] = True
    elif labels[i] == "Negative":
      negative_words[word] = True
only_negative = []
for word in negative_words:
  if word not in positive_words:
    only_negative.append(word)

print("WORDS ONLY IN NEGATIVE REVIEWS")
print(", ".join(sorted(only_negative)))
```
