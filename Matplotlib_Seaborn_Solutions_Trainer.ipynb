# 🟠 Part A: Matplotlib Solutions

### ✅ Question 1 — Line Plot

**Question:** Calculate the average marks of all students in each of the four tests. Create a line plot to show how the class average changes from test to test.

```python
tests = ["Test_1", "Test_2", "Test_3", "Test_4"]
avg_marks = df[tests].mean()
print(avg_marks.round(1))

plt.figure(figsize=(7, 4))
plt.plot(tests, avg_marks, marker="o", color="green")
plt.title("Class Average Marks Across Tests")
plt.xlabel("Test")
plt.ylabel("Average Marks")
plt.grid(True)
plt.show()
```

> 👨‍🏫 **Trainer note**
>
> **Expected observation:** The class average rises steadily from about 62.5 (Test_1) to about 72.2 (Test_4), so performance is improving.
>
> **Teaching points:** `df[tests].mean()` gives one average per column. A line plot suits ordered data (tests over time).
>
> **Common mistakes:** Plotting all 30 rows instead of the 4 averages; forgetting title/axis labels.

### ✅ Question 2 — Bar Chart

**Question:** Create a bar chart comparing the final Marks of the first five students.

```python
first5 = df.head(5)

plt.bar(first5["Student"], first5["Marks"], color="skyblue", edgecolor="black")
plt.title("Final Marks of the First Five Students")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

> 👨‍🏫 **Trainer note**
>
> **Expected values:** Asha 90, Ben 66, Charu 84, David 76, Eva 49. Asha scores highest and Eva lowest.
>
> **Teaching points:** `df.head(5)` selects the first five rows. Bars compare categories (students).
>
> **Common mistakes:** Using the whole DataFrame (30 bars); swapping x and y.

### ✅ Question 3 — Histogram

**Question:** Create a histogram of the final Marks. Change the number of bins (3, 5, 10) and observe the difference.

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, b in zip(axes, [3, 5, 10]):
    ax.hist(df["Marks"], bins=b, edgecolor="black")
    ax.set_title(f"Distribution of Marks (bins = {b})")
    ax.set_xlabel("Marks")
    ax.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```

> 👨‍🏫 **Trainer note**
>
> **Expected observation:** With few bins (3) the shape looks very coarse; with 10 bins more detail appears, but the bars get shorter and can look uneven. About 5 bins gives a balanced view. Marks range from 44 to 100, with a mean near 73.5.
>
> **Teaching points:** Bins are intervals; the number of bins changes how the same data looks. Students can also try one plot at a time with `plt.hist(df["Marks"], bins=5)`.
>
> **Common mistakes:** Confusing a histogram with a bar chart; thinking a different bin count changes the data.

### ✅ Question 4 — Scatter Plot

**Question:** Create a scatter plot showing the relationship between Study_Hours and Marks.

```python
plt.scatter(df["Study_Hours"], df["Marks"], color="crimson")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours per Day")
plt.ylabel("Final Marks")
plt.grid(True)
plt.show()

print("Correlation:", round(df["Study_Hours"].corr(df["Marks"]), 2))
```

> 👨‍🏫 **Trainer note**
>
> **Expected observation:** Yes. The points rise from bottom-left to top-right, a strong positive relationship (correlation about 0.96). Students who study more tend to score higher.
>
> **Teaching points:** Each dot is one student. Correlation does not by itself prove cause and effect.
>
> **Common mistakes:** Putting the variables on the wrong axes; using a line plot instead.

### ✅ Question 5 — Pie Chart

**Question:** Create a pie chart showing the percentage of students who prefer each programming language.

```python
lang_counts = df["Preferred_Language"].value_counts()
print(lang_counts)

plt.pie(lang_counts, labels=lang_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Preferred Programming Languages")
plt.show()
```

> 👨‍🏫 **Trainer note**
>
> **Expected values:** Python 12 students (40%), Java 8 (about 26.7%), C++ 5 (about 16.7%), Other 5 (about 16.7%). Python is the most popular.
>
> **Teaching points:** `value_counts()` converts raw rows into category counts. `autopct` prints the percentages.
>
> **Common mistakes:** Passing the raw column to `plt.pie()` instead of the counts.

# 🟢 Part B: Seaborn Solutions

### ✅ Question 6 — Count Plot

**Question:** Using Seaborn, create a count plot showing the number of students in each Class.

```python
sns.countplot(data=df, x="Class")
plt.title("Number of Students in Each Class")
plt.xlabel("Class")
plt.ylabel("Number of Students")
plt.show()
```

> 👨‍🏫 **Trainer note**
>
> **Expected result:** Both classes have 15 students, so the bars are equal.
>
> **Teaching points:** `countplot` counts rows automatically, so only `x` is needed (no `y`).
>
> **Common mistakes:** Giving both `x` and `y`; a wrong column name (names are case-sensitive).

### ✅ Question 7 — Box Plot

**Question:** Using Seaborn, create a box plot to compare the final Marks of Class A and Class B. Which class has the higher median?

```python
sns.boxplot(data=df, x="Class", y="Marks")
plt.title("Final Marks by Class")
plt.xlabel("Class")
plt.ylabel("Marks")
plt.show()

print(df.groupby("Class")["Marks"].median())
```

> 👨‍🏫 **Trainer note**
>
> **Expected observation:** Class A has the higher median (77 vs 68). Class B has the wider spread (44 to 100), while Class A ranges from 49 to 94.
>
> **Teaching points:** The line inside the box is the median; the box shows Q1 to Q3; whiskers show the spread; separate dots would be outliers.
>
> **Common mistakes:** Reading the middle line as the mean.

### ✅ Question 8 — Heatmap

**Question:** Create a heatmap of the correlation between Study_Hours, Attendance, Sleep_Hours and Marks. Which column has the strongest relationship with Marks?

```python
cols = ["Study_Hours", "Attendance", "Sleep_Hours", "Marks"]
correlation = df[cols].corr()
print(correlation.round(2))

plt.figure(figsize=(6, 5))
sns.heatmap(correlation, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

> 👨‍🏫 **Trainer note**
>
> **Expected observation:** `Study_Hours` has the strongest relationship with `Marks` (about 0.96), followed by `Attendance` (about 0.88). `Sleep_Hours` has a weak negative relationship (about -0.16), so it is almost unrelated.
>
> **Teaching points:** Values range from -1 to +1; the diagonal is always 1. Study_Hours and Attendance are also strongly related to each other (about 0.90).
>
> **Common mistakes:** Including text columns (Student, Class) in `.corr()`.

### ✅ Question 9 — Explore the Data

> 👨‍🏫 **Trainer note:** Student answers will vary. Accept any three *different* plot types from this dataset with a reasonable observation. Below are three sample answers.

**Sample Plot 1 — Bar plot: average marks by preferred language**

```python
sns.barplot(data=df, x="Preferred_Language", y="Marks", errorbar=None)
plt.title("Average Marks by Preferred Language")
plt.xlabel("Preferred Language")
plt.ylabel("Average Marks")
plt.show()
```

> **Sample observation 1:** Students who prefer Java (about 77) and C++ (about 77) have slightly higher average marks than Python (about 71) and Other (70) learners, but the differences are small.

**Sample Plot 2 — Scatter plot: sleep hours vs marks**

```python
sns.scatterplot(data=df, x="Sleep_Hours", y="Marks", hue="Class")
plt.title("Sleep Hours vs Marks")
plt.xlabel("Sleep Hours per Day")
plt.ylabel("Final Marks")
plt.show()
```

> **Sample observation 2:** There is no clear pattern. The points are scattered, so sleep hours do not strongly affect marks in this dataset (correlation about -0.16).

**Sample Plot 3 — Line plot: test averages for each class**

```python
tests = ["Test_1", "Test_2", "Test_3", "Test_4"]
class_avg = df.groupby("Class")[tests].mean()

plt.figure(figsize=(7, 4))
for c in class_avg.index:
    plt.plot(tests, class_avg.loc[c], marker="o", label=f"Class {c}")
plt.title("Average Test Marks by Class")
plt.xlabel("Test")
plt.ylabel("Average Marks")
plt.legend()
plt.grid(True)
plt.show()
```

> **Sample observation 3:** Both classes improve from Test_1 to Test_4, so the improvement is not limited to one class.
