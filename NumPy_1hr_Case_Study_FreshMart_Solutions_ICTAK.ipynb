## The scenario
**Fresh Mart** is a supermarket with **5 departments**. The manager has the daily sales of the last **4 weeks** (28 days, the first day is a **Monday**) in a NumPy array, in thousands of rupees. The data was typed by hand, so a few values are **missing** and one is **wrong**. The manager wants a clean data set, a few key insights and a neat department report.

You will use: arrays and indexing, `axis` statistics, missing values, boolean masks, `reshape`, broadcasting, sorting and formatted printing.

## The data
Run this cell first.

````python
import numpy as np
np.set_printoptions(precision=2, suppress=True)

# Sales in thousands of rupees (Rs '000): 28 days (4 weeks, starting on a Monday) x 5 departments
departments = np.array(["Grocery", "Dairy", "Bakery", "Fruits", "Snacks"])
days_of_week = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
sales = np.array([
    [49.5, 30.4, 16.0, 28.2, 22.8],
    [48.5, 28.0, 17.4, 24.3, 20.6],
    [52.6, 30.0, 17.4, 25.1, 21.5],
    [50.1, np.nan, 18.6, 25.8, 20.2],
    [55.6, 34.3, 19.5, 28.3, 25.1],
    [77.9, 38.8, 26.3, 32.5, 30.0],
    [62.9, 42.2, 24.6, 36.1, 27.1],
    [55.8, 29.9, 18.0, 27.6, 23.9],
    [51.5, 28.2, 16.7, 27.6, 22.3],
    [54.2, 33.9, 17.1, 30.0, 26.1],
    [58.8, 32.4, 17.8, np.nan, 22.1],
    [58.8, 33.4, 20.7, 31.7, 24.1],
    [68.0, 39.2, -18.0, 33.0, 28.9],
    [75.0, 38.7, 24.5, 32.3, 29.8],
    [49.4, 30.5, 21.7, 26.6, 24.9],
    [53.8, 29.9, 17.4, 28.2, 21.9],
    [58.4, 29.5, 18.6, 27.5, 23.1],
    [np.nan, 30.5, 18.9, 26.1, 24.3],
    [62.8, 30.9, 18.7, 28.0, 25.8],
    [83.5, 41.9, 26.1, 34.8, 29.5],
    [63.2, 41.3, 26.2, 35.1, 32.9],
    [57.4, 32.6, 20.2, 29.2, 25.6],
    [51.3, 26.8, 18.8, 25.9, np.nan],
    [60.5, 29.6, 19.9, 29.0, 23.9],
    [60.5, 31.8, 18.8, 30.1, 25.0],
    [67.5, 36.4, 23.7, 30.6, 24.7],
    [67.2, 41.2, 27.8, 34.9, 31.5],
    [83.0, 44.7, 28.2, 34.7, 26.6]
])

# Weekly sales target for each department (Rs '000)
weekly_target = np.array([420, 235, 140, 200, 170])
````

## Stage A · Warm-up (beginner)

### Task 1 · Know your data
Print the shape, the number of values and the dtype of `sales`, one per line.

**Expected output**
```
(28, 5)
140
float64
```

````python
print(sales.shape)
print(sales.size)
print(sales.dtype)
````

### Task 2 · Pick values
Print the sales of all departments on day 1. Then print the Bakery sales (column 2) for the first 7 days.

**Expected output**
```
[49.5 30.4 16.  28.2 22.8]
[16.  17.4 17.4 18.6 19.5 26.3 24.6]
```

````python
print(sales[0])
print(sales[:7, 2])
````

### Task 3 · Department totals
Some values are missing (`nan`), so first print how many values are missing in the **whole** array. Then print the total sales of each department using a function that **ignores** the missing values (1 decimal).

**Expected output**
```
4
[1637.7  917.   541.6  803.2  684.2]
```

````python
print(np.isnan(sales).sum())
print(np.round(np.nansum(sales, axis=0), 1))
````

### Task 4 · Lowest value
Print the smallest value in the whole array (ignoring missing values) and the day number (1 to 28) and department name where it occurs.

**Expected output**
```
-18.0
13 Bakery
```

````python
low = np.nanmin(sales)
row, col = np.argwhere(sales == low)[0]
print(low)
print(row + 1, departments[col])
````

## Stage B · Clean the data

### Task 5 · Missing values by department
Print how many values are missing in **each department** (one number per department).

**Expected output**
```
[1 1 0 1 1]
```

````python
print(np.isnan(sales).sum(axis=0))
````

### Task 6 · Fill the gaps
Create a copy `clean` of `sales` in which every missing value is replaced by the **average of its own department** (ignoring missing values). Print the number of missing values left and the department averages used (1 decimal).

**Expected output**
```
0
[60.7 34.  19.3 29.7 25.3]
```

````python
dept_mean = np.nanmean(sales, axis=0)
clean = np.where(np.isnan(sales), dept_mean, sales)
print(np.isnan(clean).sum())
print(np.round(dept_mean, 1))
````

### Task 7 · Impossible sales
Sales can never be negative: a negative value is a refund typed by mistake. Print how many negative values are in `clean`, then replace them with 0 (in `clean`) and print the smallest value of `clean`.

**Expected output**
```
1
0.0
```

````python
print((clean < 0).sum())
clean[clean < 0] = 0
print(clean.min())
````

## Stage C · Analyse (intermediate)

### Task 8 · Weeks view
Reshape `clean` into a 3-D array `weekly` of shape (4 weeks, 7 days, 5 departments) and print its shape. Then print the total sales of every department in every week as a 4 × 5 table (1 decimal).

**Expected output**
```
(4, 7, 5)
[[397.1 237.7 139.8 200.3 167.3]
 [422.1 235.7 114.8 211.9 177.2]
 [431.8 234.5 147.6 206.3 182.4]
 [447.4 243.1 157.4 214.4 182.6]]
```

````python
weekly = clean.reshape(4, 7, 5)
print(weekly.shape)
week_totals = weekly.sum(axis=1)
print(np.round(week_totals, 1))
````

### Task 9 · Best day of the week
Add the departments together to get the sales of the whole store for each of the 28 days. Then find the **average store sales for each weekday** (Mon to Sun) across the 4 weeks. Print the 7 averages (1 decimal) and the name of the best weekday.

**Expected output**
```
[155.  146.1 157.  159.  170.2 198.3 202.3]
Sun
```

````python
store_daily = clean.sum(axis=1)
weekday_avg = store_daily.reshape(4, 7).mean(axis=0)
print(np.round(weekday_avg, 1))
print(days_of_week[weekday_avg.argmax()])
````

### Task 10 · Weekend or weekday?
Saturday and Sunday are the weekend. Print the **average daily sales of each department** on weekdays (Mon to Fri) and on weekends, as two lines (1 decimal).

**Expected output**
```
[55.9 31.1 18.8 28.  23.7]
[72.6 41.  23.  34.2 29.5]
```

````python
is_weekend = np.tile(np.arange(7) >= 5, 4)
print(np.round(clean[~is_weekend].mean(axis=0), 1))
print(np.round(clean[is_weekend].mean(axis=0), 1))
````

### Task 11 · Growth
For each department, find the **percentage growth** from the week 1 total to the week 4 total. Print the 5 values (1 decimal) and the name of the fastest-growing department.

**Expected output**
```
[12.7  2.3 12.6  7.   9.2]
Grocery
```

````python
growth = (week_totals[3] - week_totals[0]) / week_totals[0] * 100
print(np.round(growth, 1))
print(departments[growth.argmax()])
````

### Task 12 · Busy days
A day is **busy** if the store sales for that day are more than **1.2 times** the average daily store sales. Print the number of busy days and their day numbers (1 to 28).

**Expected output**
```
3
[ 6 20 28]
```

````python
busy = store_daily > 1.2 * store_daily.mean()
print(busy.sum())
print(np.where(busy)[0] + 1)
````

## Stage D · Decide (intermediate)

### Task 13 · Targets met?
Compare the 4 × 5 weekly totals with `weekly_target` (a department's target is the same every week). Print the True/False table, then the number of weeks in which each department met its target (total at least the target).

**Expected output**
```
[[False  True False  True False]
 [ True  True False  True  True]
 [ True False  True  True  True]
 [ True  True  True  True  True]]
[3 3 2 4 3]
```

````python
met = week_totals >= weekly_target
print(met)
print(met.sum(axis=0))
````

### Task 14 · Department ranking
Rank the departments from the highest to the lowest 4-week total. Print their names in that order, and then the **share of total store sales** of each department in that order (percentage, 1 decimal).

**Expected output**
```
['Grocery' 'Dairy' 'Fruits' 'Snacks' 'Bakery']
[35.7 20.  17.5 14.9 11.8]
```

````python
dept_total = clean.sum(axis=0)
order = dept_total.argsort()[::-1]
print(departments[order])
print(np.round(dept_total[order] / dept_total.sum() * 100, 1))
````

---
## Stage E · Report

### Task 15 · Print the report
Print a table with the headings `Department`, `Total` and `Share %` (Department left-aligned in 12 characters, Total and Share % right-aligned in 10 each), a line of 32 dashes, and one row per department in ranked order (total with 1 decimal, share with 1 decimal).

**Expected output**
```
Department       Total   Share %
--------------------------------
Grocery         1698.4      35.7
Dairy            951.0      20.0
Fruits           832.9      17.5
Snacks           709.5      14.9
Bakery           559.6      11.8
```

````python
print(f"{'Department':<12}{'Total':>10}{'Share %':>10}")
print("-" * 32)
for i in order:
    share = dept_total[i] / dept_total.sum() * 100
    print(f"{departments[i]:<12}{dept_total[i]:>10.1f}{share:>10.1f}")
````

## Stage Challenge (optional)

### Task 16 · 7-day moving average
A **moving average** smooths daily numbers. For `store_daily`, calculate the average of every 7 consecutive days (days 1–7, days 2–8, and so on). Print how many averages there are and the first 3 (1 decimal).

**Expected output**
```
22
[163.2 164.4 165.4]
```

````python
ma = np.convolve(store_daily, np.ones(7) / 7, mode="valid")
print(len(ma))
print(np.round(ma[:3], 1))
````
